# Kitchen Receipts: train the checker

The checker is a small model (decision trees, no GPU needed) that learns from **your own checks**: for every value of
every invoice you checked and saved, the phone noted how it was read (flagged or not, where it came from, whether the
document's sums proved it) and whether you kept it or corrected it. **Numbers only**: the file holds no supplier, number,
amount or text of your documents.

It learns which flagged values are in fact right, so that later the app can ask you fewer of them. It is **ready** only
when, on invoices it did not learn from, it would have cleared **no wrong value** and let **no invoice with a wrong
value** save by itself. Until then nothing changes in the app.

**Steps:** in the app, Settings ▸ *Teach the checker* ▸ *Send the checker data* (to yourself, e.g. your Google Drive).
Then run the cells below (Runtime ▸ Run all) and upload that zip when asked. A few hundred checked invoices are needed.
